```python
df = df.drop_duplicates()
```

```python
df["Description"] = df["Description"].fillna("")
```

```python
df["Priority"] = df["Priority"].fillna("Unknown")
df["Category"] = df["Category"].fillna("Unknown")
df["Assigned_Team"] = df["Assigned_Team"].fillna("Unknown")
df["Status"] = df["Status"].fillna("Unknown")
```

```python
df["SLA_Breached"] = df["SLA_Breached"].replace({
    "Y": "Yes",
    "N": "No",
    "FALSE": "No",
    "breach": "Yes"
})

df["SLA_Breached"] = df["SLA_Breached"].fillna("Unknown")
```

```python
print(df.isnull().sum())
```

```python
df = df.drop_duplicates()

df.to_csv("it_support_cleaned.csv", index=False)
```

```python
df["Description"] = df["Description"].str.lower()
```

```python
print(df["Description"].head())
```

Remove special characters


```python
import re

df["Description"] = df["Description"].apply(
    lambda x: re.sub(r"[^a-zA-Z0-9\s]", "", x)
)
```

Remove extra spaces 


```python
df["Description"] = df["Description"].str.strip().str.replace(r"\s+", " ", regex=True)
```

```python
df.to_csv("it_support_preprocessed.csv", index=False)

print("Dataset saved successfully!")
```
