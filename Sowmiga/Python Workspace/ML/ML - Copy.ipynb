```python

import pandas as pd
from sklearn.preprocessing import LabelEncoder,MinMaxScaler,StandardScaler
import seaborn as sns
import numpy as np

```

```python
df=pd.read_excel(r"C:\Users\ELCOT\Desktop\Sowmiga\Python Workspace\ML\Practice_data.xlsx")
```

```python
df.sample(5)
```

```python
x=df.drop(["Final_Score"],axis=1)
y=df["Final_Score"]
```

```python
X = df[
    ["Attendance_Pct",
     "Study_Hours_Per_Day","Internal_Marks","Assignments_Completed"
    ]
]

y = df["Final_Score"]
```

```python
from sklearn.model_selection import train_test_split

# Optional but recommended: fill missing values before modeling
X = X.fillna(X.mean(numeric_only=True))

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report
from sklearn.tree import DecisionTreeClassifier
models={ "logistic regression": LogisticRegression(max_iter=1000),
        "decision tree": DecisionTreeClassifier( max_depth=5,random_state=42),
        "random forest": RandomForestClassifier(n_estimators=100, random_state=42),}
for name,model in models.items():
    model.fit(X_train,y_train)
    y_pred=model.predict(X_test)
    print(f"\n{name}")
    print(classification_report(y_test,y_pred,zero_division=0))
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report
from sklearn.tree import DecisionTreeClassifier
df=pd.read_excel(r"C:\Users\ELCOT\Desktop\Sowmiga\Python Workspace\ML\Practice_data_with_pass_fail.xlsx")
X=df[["Attendance_Pct", "Study_Hours_Per_Day", "Internal_Marks", "Assignments_Completed"]]
y = df["pass_fail"].map({
    "Pass": 1,
    "Fail": 0
})
X_test,Y_test,X_train,Y_train=train_test_split(X,Y,test_size=0.3,random_state=42)
models={ "logistic regression": LogisticRegression(max_iter=1000),
        "Decision tree": DecisionTreeClassifier(max_depth=5,random_state=42),
        "random forest": RandomForestClassifier(n_estimators=100, random_state=42)}
for name,model in models.items():
    model.fit (X_train,Y_train)
    y_pred=model.predict(X_test)
    print(f"\n{name}")
    print(classification_report(Y_test,y_pred,zero_division=0))
```
