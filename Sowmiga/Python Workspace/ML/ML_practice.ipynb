Logistic Regression - ML data(sheet2)

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score,confusion_matrix, precision_score, recall_score, classification_report
```

```python
# Fix for the invalid sheet index
data = pd.read_excel(r"C:\Users\ELCOT\Desktop\Sowmiga\Python Workspace\ML\ML_data.xlsx", sheet_name=0)
```

```python
df=pd.DataFrame(data)
df
```

```python
df["Pass_Fail"] = df["Final_Result"].map({"Pass": 1, "Fail": 0})
df["Internal_Marks"] = df["Internal_Marks"].fillna(df["Internal_Marks"].median())
df = pd.get_dummies(df, columns=["Department", "Gender"], drop_first=True)

X = df[["Attendance", "Study_Hours", "Internal_Marks", "Backlogs",
    "Department_IT", "Department_ECE", "Gender_Male"]]
y = df["Pass_Fail"]
y=df["Pass_Fail"]
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state=42)

model_logistic=LogisticRegression()

# Ensure both classes are present in train and test sets
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42, stratify=y
)

model_logistic.fit(X_train, y_train)

y_pred_logistic=model_logistic.predict(X_test)

#metrics
#accuracy
acc=accuracy_score(y_test,y_pred_logistic)

#Confusion matrix
con_mat=confusion_matrix(y_test,y_pred_logistic)

# precision
precision_logistic = precision_score(y_test, y_pred_logistic, pos_label=1)  # labels are 0/1, not "Pass"/"Fail"



# pos_label means positive label. It tells Python which class you want to treat as the positive class
# when calculating precision, recall, or F1-score for binary classification

# recall
recall_logistic = recall_score(y_test, y_pred_logistic, pos_label=1)

# classification report
report=classification_report(y_test,y_pred_logistic)

# Support - in actual data , the count of actual pass and fail data of test data


print(acc)
print(50*"-")
print(y_pred_logistic)
print(50*"-")
print(y_test)
print(50*"-")
print(con_mat)
print(50*"-")
print("Precision")
print(precision_logistic)
print(50*"-")
print("recall")
print(recall_logistic)
print(50*"-")
print(report)
```
